```python
import os
from sqlalchemy import create_engine, text
from dotenv import load_dotenv
import pandas as pd

def connect_to_database():
    DATABASE_URL = (
        "postgresql://postgres:1234"
        "@localhost:5432/google_and_spotify"
    )

    engine = create_engine(
        DATABASE_URL
    )
    return engine
```

```python
engine = connect_to_database()

with engine.connect() as conn:
    result = conn.execute(text("SELECT 1"))
    print(result.scalar())
```

```python
df_tracks = pd.read_sql(
    "SELECT * FROM track_audio_features",
    engine
)
df_tracks.head()
```

```python
df_tracks.shape
```

```python
df_fact_listening = pd.read_sql(
    "SELECT * FROM fact_listening",
    engine
)
df_fact_listening.head()
```

```python
df_fact_listening.shape
```

```python
df_tracks_fact_listening = df_fact_listening.merge(
    df_tracks,
    on="track_id",
    how="left",
    suffixes=("_listening", "_track")
)

df_tracks_fact_listening.head()
```

```python
audio_features = [
    "acousticness",
    "danceability",
    "energy",
    "instrumentalness",
    "key",
    "liveness",
    "loudness",
    "mode",
    "speechiness",
    "tempo",
    "valence",
    "popularity",
]

# base de features por track
tracks_model = (
    df_tracks
    .drop_duplicates(subset=["track_id"])
    .loc[:, ["track_id"] + audio_features]
    .fillna(0)
)

# quantidade de streams por track
streams_por_track = (
    df_fact_listening
    .groupby("track_id", as_index=False)
    .agg(quantidade_streams=("listening_id", "count"))
)

# dataframe final para treino
# keep: track_id, quantidade_streams + audio features
# opcional: target binário para gostar/nao gostar
# gosto = 1 quando a música teve pelo menos 1 stream

df_model = tracks_model.merge(streams_por_track, on="track_id", how="left")

df_model["quantidade_streams"] = df_model["quantidade_streams"].fillna(0).astype(int)
df_model["gostei"] = (df_model["quantidade_streams"] > 0).astype(int)

# ordem final das colunas
cols = ["track_id", "quantidade_streams"] + audio_features + ["gostei"]
df_model = df_model[cols]

df_model.head()
```

```python
df_model
```

```python
import numpy as np

df_model["gostei"] = np.where(
    df_model["quantidade_streams"] > 10,
    1,
    0
)
```

```python
df_model
```

```python
artist_track = pd.read_sql(
    "SELECT track_id, artist_name FROM dim_track",
    engine
)

df_model = df_model.merge(
    artist_track,
    on="track_id",
    how="left"
)

df_model = df_model[
    ["track_id", "artist_name", "quantidade_streams"] + audio_features + ["gostei"]
]

df_model.head()
```

```python
df_model.value_counts("gostei")
```

```python
df_model = df_model.drop("quantidade_streams", axis=1)
```

```python
from sklearn.model_selection import train_test_split

X = df_model.drop("gostei", axis=1)
y = df_model["gostei"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
y_test.value_counts()
```

```python
from catboost import CatBoostClassifier

model = CatBoostClassifier(
    iterations=500,
    learning_rate=0.05,
    depth=6,
    random_seed=42,
    verbose=False
)

model.fit(X_train, y_train, cat_features=["artist_name"])
```

```python
model.score(X_test, y_test)
```

```python
y_test
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

y_pred = model.predict(X_test).flatten()

print(f"Accuracy:  {accuracy_score(y_test, y_pred):.4f}")
print(f"Precision: {precision_score(y_test, y_pred):.4f}")
print(f"Recall:    {recall_score(y_test, y_pred):.4f}")
print(f"F1 Score:  {f1_score(y_test, y_pred):.4f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
y_proba = model.predict_proba(X_test)[:, 1]

for threshold in [0.3, 0.35, 0.4, 0.45, 0.5]:
    y_pred = (y_proba >= threshold).astype(int)

    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)

    print(
        f"Threshold={threshold:.2f} | "
        f"Precision={precision:.3f} | "
        f"Recall={recall:.3f} | "
        f"F1={f1:.3f}"
    )
```
