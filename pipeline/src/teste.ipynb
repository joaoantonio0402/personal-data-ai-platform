```python
import requests


def search_track(track_name, artist_name):
    url = "https://api.reccobeats.com/v1/track/search"

    params = {
        "searchText": track_name
    }

    response = requests.get(url, params=params)
    response.raise_for_status()

    data = response.json()

    track_name_normalized = track_name.strip().lower()
    artist_name_normalized = artist_name.strip().lower()

    for track in data.get("content", []):
        title = track.get("trackTitle", "").strip().lower()

        artists = [
            artist.get("name", "").strip().lower()
            for artist in track.get("artists", [])
        ]

        if (
            title == track_name_normalized
            and artist_name_normalized in artists
        ):
            return track

    return None
```

```python
track = search_track("Hair Down", "SiR")

if track:
    print("Encontrada!")
    print(track["id"])
    print(track["trackTitle"])
    print([artist["name"] for artist in track["artists"]])
else:
    print("Música não encontrada.")
```
